```python
df = Student.copy()
df["Marks_pct"] = df["Marks"] / 100
df
```

```python
df["passed"] = df["Marks"] >= 70
df
```

```python
df["name_upper"] = df["Name"].str.upper()
df
```

```python
df["name_lower"] = df["Name"].str.lower()
df
```

```python
df["Grade"] = pd.cut (df["Marks"],
                      bins = [0,69,79,89,100],
                      labels = ["D", "C", "B", "A"])
df
```

```python
print("missing per column : \n",df.isna().sum())
```
